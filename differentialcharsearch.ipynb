```python
from aradi import *
from printing import *
```

```python
r = 7
offset = 2

model = cp_model.CpModel()
cor_vars = []

differences = [[model.NewBoolVar("") for _ in range(128)]  for _ in range(r+1)]
# [[0]*96 + [model.NewBoolVar("") for _ in range(32)] for _ in range(r+1)]+ [[model.NewBoolVar("") for _ in range(128)]] 
for i in range(r):
    model, ecv = aradi_differential_trails_round(model, differences[i], differences[i+1],i+offset)
    cor_vars += ecv

model.Add(sum(differences[0])==1)
model.Add(sum(differences[r])==1)

model.minimize(sum(cor_vars))

solver = cp_model.CpSolver()
solver.parameters.max_time_in_seconds = 600.0 #search for a set time
status = solver.Solve(model)
if status == cp_model.OPTIMAL:
    print("optimal solution found", solver.ObjectiveValue())
    print_trail3(differences, solver)
    print_aradi_trail(differences, solver)
elif status == cp_model.FEASIBLE:
    print("non-optimal solution found" ,solver.ObjectiveValue())
    print_trail3(differences, solver)
    print_aradi_trail(differences, solver)
elif status == cp_model.INFEASIBLE:
    print("no solution found")
else:
    print("stopped")
```
